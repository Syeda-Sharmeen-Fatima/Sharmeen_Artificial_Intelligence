```python
def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for karachi tourism.

Provide a well-structured travel plan of Karachi with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response small and easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model="models/gemini-3.7-flash",
        contents=prompt
    )

    return response.text



print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
from google import genai
```

```python
client = genai.Client(api_key="YOUR_API_KEY")
```

```python
# Student Data
student_data = {
    "bilal": {
        "math": 85,
        "english": 78,
        "physics": 90
    },
    "affan": {
        "math": 72,
        "english": 88,
        "physics": 80
    },
    "zainab": {
        "math": 95,
        "english": 91,
        "physics": 87
    }
}


# Gemini fallback function
def student_agent(user_input):

    prompt = f"""
You are a student marks assistant.

Generate a reasonable example mark between 0 and 100.

Return ONLY the number of marks.

User request:
{user_input}
"""

    response = client.models.generate_content(
        model="models/gemini-3.7-flash",
        contents=prompt
    )

    return response.text.strip()


# Main Program
print("Welcome to Student Marks Agent (type 'exit' to stop)\n")

while True:

    name = input("Enter student name: ").strip().lower()

    if name == "exit":
        print("Goodbye!")
        break

    subject = input("Enter subject: ").strip().lower()

    if subject == "exit":
        print("Goodbye!")
        break

    try:

        if name in student_data:

          if subject in student_data[name]:

                # Marks found in dictionary
                marks = student_data[name][subject]

          else:

                # Subject not found, ask Gemini
                user_input = f"""
Student: {name}
Subject: {subject}
"""

                marks = student_agent(user_input)

        else:

            # Student not found, ask Gemini
            user_input = f"""
Student: {name}
Subject: {subject}
"""

            marks = student_agent(user_input)

        display_name = name.title()
        display_subject = subject.title()

        print(f"\nAgent: {display_name} got {marks} marks in {display_subject}.\n")

    except Exception as e:

      print("Error:", e)
```
